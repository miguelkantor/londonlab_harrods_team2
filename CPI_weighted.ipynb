```python
import pandas as pd
cpi = pd.read_csv('CPI.csv')  
```

```python
cpi
```

```python
# 转换为长格式
cpi_long = pd.melt(cpi,
                  id_vars=['COUNTRY', 'COUNTRY_CODE', 'REGION'],  # 保持不变的列
                  var_name='MONTH',       # 新的列名（原来的列名如2022-M01）
                  value_name='VALUE')     # 新的数值列

cpi_long = cpi_long.sort_values(['COUNTRY', 'MONTH'])  # 确保顺序
cpi_long
```

```python
cpi_long['VALUE'] = cpi_long.groupby('COUNTRY')['VALUE'].ffill().bfill()

# 保存为新的文件
cpi_long.to_csv('CPI_long_format.csv', index=False)

cpi_long
```

```python
df_cpi= pd.read_csv('CPI_long_format.csv') 
```

```python
df_cpi['YEAR'] = df_cpi['MONTH'].str.extract(r'(\d{4})').astype(int)
df_cpi
```

```python
df_gdp = pd.read_csv('GDP.csv')
df_gdp = pd.melt(df_gdp,
                      id_vars=['COUNTRY', 'COUNTRY_CODE', 'REGION'],
                      var_name='YEAR',
                      value_name='GDP')
df_gdp['YEAR'] = df_gdp['YEAR'].astype(int)

df_gdp = df_gdp.sort_values(['COUNTRY', 'YEAR'])

df_gdp['GDP'] = df_gdp.groupby('COUNTRY')['GDP'].ffill().bfill()
df_gdp
```

```python
df_merged_cpi = pd.merge(df_cpi, df_gdp, on=['COUNTRY','COUNTRY_CODE', 'REGION', 'YEAR'], how='inner')
df_merged_cpi
```

```python
df_merged_cpi['GDP_WEIGHTED_CPI'] = df_merged_cpi['VALUE'] * df_merged_cpi['GDP']
df_merged_cpi
```

```python
df_cpi_result = (
    df_merged_cpi
    .groupby(['REGION', 'MONTH'])
    .apply(lambda g: g['GDP_WEIGHTED_CPI'].sum() / g['GDP'].sum())
    .reset_index(name='REGION_CPI_WEIGHTED')
)
df_cpi_result
```

```python
df_cpi_result['MONTH'] = df_cpi_result['MONTH'].str.replace('-M', '/')
df_cpi_result.to_csv('region_weighted_cpi.csv', index=False)
```
