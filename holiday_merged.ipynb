# Visualising

```python
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.dates import MonthLocator, DateFormatter

def plot_facet_lines(
    df,
    countries,
    date_col,
    y1_col,
    y2_col,
    y1_label=None,
    y2_label=None,
    col_wrap=3,
    height=5,
    aspect=1.5
):
    """
    Faceted line plots of two series over time, one on primary y-axis (blue),
    one on secondary y-axis (green), for specified countries, with a global legend.
    """
    y1_label = y1_label or y1_col
    y2_label = y2_label or y2_col

    filtered = df[df['COUNTRY'].isin(countries)]

    # Ensure date column is in datetime format
    filtered.loc[:, date_col] = pd.to_datetime(filtered[date_col])

    sns.set(style='whitegrid')
    g = sns.FacetGrid(
        filtered,
        col='COUNTRY',
        col_wrap=col_wrap,
        height=height,
        aspect=aspect,
        sharey=False
    )

    def _plot(data, **kwargs):
        ax = plt.gca()
        # Sort data by date
        data = data.sort_values(by=date_col)
        
        # primary series (blue)
        ax.plot(data[date_col], data[y1_col], color='tab:blue')
        ax.set_ylim(bottom=0)
        ax.set_ylabel(y1_label)
          
        # Format x-axis with one tick per month
        ax.xaxis.set_major_locator(MonthLocator(interval=3))
        ax.xaxis.set_major_formatter(DateFormatter('%Y-%m'))
        ax.tick_params(axis='x', rotation=45, labelsize=8)
        
        # secondary series (green)
        ax2 = ax.twinx()
        ax2.plot(data[date_col], data[y2_col], color='tab:green')
        ax2.set_ylim(bottom=0)
        ax2.set_ylabel(y2_label)
        # x-labels
        ax.tick_params(axis='x', rotation=30, labelsize=8)

    g.map_dataframe(_plot)

    # Create a global legend
    blue_line = plt.Line2D([], [], color='tab:blue', label=y1_label)
    green_line = plt.Line2D([], [], color='tab:green', label=y2_label)
    # place legend at top right INSIDE the figure
    g.fig.legend(
        handles=[blue_line, green_line],
        loc='upper right',
        bbox_to_anchor=(0.95, 0.95),
        frameon=False
    )

    plt.tight_layout()
    plt.show()
```

```python
plot_facet_lines(
    merged_week_country,
    countries=['Middle East - GCC'],
    date_col='DATE',
    y1_col='ARRIVALS',
    y2_col='TRANSACTIONS',
    y1_label='Arrivals',
    y2_label='Transactions',
    col_wrap=3
)
```

# Adding Holidays



```python
import pandas as pd
from datetime import timedelta

# Load the full dataset
df = pd.read_csv("merged_week_country.csv", parse_dates=["DATE"])

# Define holiday periods (Eid al-Fitr, Eid al-Adha, National Day)
holiday_ranges = [
    # Eid al-Fitr
    ("2022-05-01", "2022-05-07"),
    ("2023-04-20", "2023-04-27"),
    ("2024-04-09", "2024-04-15"),
    ("2025-03-30", "2025-04-05"),

    # Eid al-Adha
    ("2022-07-08", "2022-07-13"),
    ("2023-06-27", "2023-07-02"),
    ("2024-06-14", "2024-06-20"),
    ("2025-06-04", "2025-06-10"),

    # Saudi National Day
    ("2022-09-23", "2022-09-23"),
    ("2023-09-23", "2023-09-23"),
    ("2024-09-23", "2024-09-23")
]

# Convert ranges to datetime tuples
holiday_windows = [(pd.to_datetime(start), pd.to_datetime(end)) for start, end in holiday_ranges]

# Function to determine if a date is during a holiday week
def is_holiday(date):
    for start, end in holiday_windows:
        if date <= end and (date + timedelta(days=6)) >= start:
            return 1
    return 0

# Filter to only Middle East - GCC and apply holiday flag
middle_east_gcc_w_holiday = df[df["COUNTRY"] == "Middle East - GCC"].copy()
middle_east_gcc_w_holiday["holiday"] = middle_east_gcc_w_holiday["DATE"].apply(is_holiday)

# Save to CSV
middle_east_gcc_w_holiday.to_csv("/Users/isisommerfeld/Desktop/LondonLAB/middle_east_gcc_w_holiday.csv", index=False)

```

# Facetting by year


```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.dates import MonthLocator, DateFormatter

# Load your dataset (adjust path as needed)
df = pd.read_csv("merged_week_country.csv", parse_dates=["DATE"])

# Filter for Middle East - GCC
df = df[df["COUNTRY"] == "Middle East - GCC"].copy()
df["YEAR"] = df["DATE"].dt.year

# Plot settings
years = [2022, 2023, 2024, 2025]

for year in years:
    yearly_data = df[df["YEAR"] == year].sort_values(by="DATE")
    
    if yearly_data.empty:
        continue  # skip empty years

    fig, ax1 = plt.subplots(figsize=(12, 5))
    
    # Primary axis: Arrivals
    ax1.plot(yearly_data["DATE"], yearly_data["ARRIVALS"], label="Arrivals", color="tab:blue")
    ax1.set_ylabel("Arrivals", color="tab:blue")
    ax1.tick_params(axis="y", labelcolor="tab:blue")
    
    # Secondary axis: Transactions
    ax2 = ax1.twinx()
    ax2.plot(yearly_data["DATE"], yearly_data["TRANSACTIONS"], label="Transactions", color="tab:green")
    ax2.set_ylabel("Transactions", color="tab:green")
    ax2.tick_params(axis="y", labelcolor="tab:green")
    
    # X-axis formatting
    ax1.set_xlim(pd.Timestamp(f"{year}-01-01"), pd.Timestamp(f"{year}-12-31"))
    ax1.xaxis.set_major_locator(MonthLocator(interval=1))
    ax1.xaxis.set_major_formatter(DateFormatter("%b"))
    ax1.tick_params(axis='x', rotation=45)

    # Title and layout
    plt.title(f"Middle East - GCC: Arrivals and Transactions in {year}")
    fig.tight_layout()
    plt.show()
```
